```python
import numpy as np
 
# Entradas
x = np.array([1, 2])
 
w1= np.array([[0.5, -0.4],
	     [0.3, 0.8],
	     [-0.6, -0.2]])
 
 
# Bias da primeira camada oculta
b1 = np.array([0.1, 0.2, 0.05])
 
# Função de ativação (ReLU)
def relu(z):
	return np.maximum(0, z)
 
# Cálculo para a camada oculta 1
z1 = w1 @ x + b1
a1 = relu(z1)
 
# Pesos e bias da camada de saída
w2 = np.array([0.7, -0.5, 0.9])
b2 = 0.2
 
# Cálculo para a camada oculta 2
z2 = w2 @ a1 + b2

# Função de ativação sigmoide (classificação)
def sigmoide(z) :
	return 1 / (1 + np.exp(-z))
 
# Saída final
y_regressao = z2 
y_classificacao = sigmoide(z2)
 
# Exibir resultados
print(f"Saída para regressão: {y_regressao}")
print(f"Saída para classificação: {y_classificacao}")
```
